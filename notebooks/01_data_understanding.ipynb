## 1. Load the tables

```python
import pandas as pd

path = '../data/'   # the CSV files are in the data folder, one level up

df_customers   = pd.read_csv(path + 'olist_customers_dataset.csv')
df_geolocation = pd.read_csv(path + 'olist_geolocation_dataset.csv')
df_items       = pd.read_csv(path + 'olist_order_items_dataset.csv')
df_payments    = pd.read_csv(path + 'olist_order_payments_dataset.csv')
df_reviews     = pd.read_csv(path + 'olist_order_reviews_dataset.csv')
df_orders      = pd.read_csv(path + 'olist_orders_dataset.csv')
df_products    = pd.read_csv(path + 'olist_products_dataset.csv')
df_sellers     = pd.read_csv(path + 'olist_sellers_dataset.csv')
df_category    = pd.read_csv(path + 'product_category_name_translation.csv')
```

```python
tables = {
    'customers': df_customers, 'geolocation': df_geolocation, 'items': df_items,
    'payments': df_payments, 'reviews': df_reviews, 'orders': df_orders,
    'products': df_products, 'sellers': df_sellers, 'category': df_category,
}
for name, df in tables.items():
    print(f'{name:12} rows: {df.shape[0]:>8}  columns: {df.shape[1]}')
```

## 2. Missing values

```python
for name, df in tables.items():
    nulls = df.isnull().sum()
    nulls = nulls[nulls > 0]
    print(f'--- {name} ---')
    if nulls.empty:
        print('no missing values')
    else:
        print(nulls)
    print()
```

## 5. Duplicates

```python
for name, df in tables.items():
    duplicates = df.duplicated().sum()
    print(f'--- {name} ---')
    if duplicates == 0:
        print('no duplicate rows')
    else:
        print(duplicates)
    print()
```

**Result:** only `geolocation` has duplicate rows (261,831). I do not use this table, so this is not a problem.

```python
df_orders.duplicated(subset='order_id').sum()
```

`order_id` is unique in `orders`: no order appears twice.

## 6. First cleaning before loading

- `has_comment`: True if the customer wrote a comment
- Products without a category get the category `unknown`
- `df_delivered`: only delivered orders with a delivery date (96,470 orders). In the final table this filter is done in SQL.

```python
# 1. reviews: does the review have a written comment?
df_reviews['has_comment'] = df_reviews['review_comment_message'].notnull()
# 2. products: empty category -> 'unknown'
df_products['product_category_name'] = df_products['product_category_name'].fillna('unknown')

# 3. orders: only delivered orders with a delivery date
df_delivered = df_orders[
    (df_orders['order_status'] == 'delivered') &
    (df_orders['order_delivered_customer_date'].notnull())
].copy()

print(len(df_orders), '->', len(df_delivered))
```

## 7. Summary

**Missing values**
- `reviews`: most customers did not write a comment title (87,656 missing) or message (58,247 missing). `review_score` is always there.
- `orders`: some delivery dates are missing (`order_delivered_customer_date`: 2,965). These are mostly orders that were not delivered.
- `products`: 610 products have no category. I filled them with `unknown` (step 6).
- The other tables have no missing values.

**Data types**
- The date columns were read as text. I converted the order dates to datetime.
- Review dates were not converted here.

**Duplicates**
- Only `geolocation` has duplicate rows. I do not use this table.

**Things to watch in the next steps**
- `order_items` has several rows per order (112,650 rows vs 99,441 orders). Joining it can create several rows for one order.
- `geolocation` has many rows per zip code, so I do not join it.

**Next step:** load the tables into MySQL and build one analysis table there.

## 8. Load the tables into MySQL

The tables are saved in the MySQL database `olist`, so the joins can be done in SQL (`sql/01_build_analysis_table.sql`). The password is asked when the cell runs, so it is not saved in the notebook. `geolocation` is not loaded, because it is not needed.

```python
import sqlalchemy
import pymysql
print(sqlalchemy.__version__)
```

```python
import getpass
from sqlalchemy import create_engine

password = getpass.getpass('MySQL password: ')
engine = create_engine(f'mysql+pymysql://root:{password}@localhost:3306/olist')

engine.connect()
print('connected')
```

```python
df_category.to_sql('product_category_name_translation', engine, index=False, if_exists='replace')
print('done')
```

```python
for name, df in tables.items():
    if name in ['geolocation', 'category']:
        continue
    df.to_sql(name, engine, index=False, if_exists='replace')
    print(name, 'done')
```
